## Importing required libraries

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn import tree
from sklearn.preprocessing import LabelEncoder
import pandas as pd
import matplotlib.pyplot as plt
```

## Loading dataset

```python
df = pd.read_csv('play_tennis_dataset.csv')
df.head()
```

## Spliting dataset

```python
attributes = df.drop('PlayTennis', axis=1)
result = df['PlayTennis']
```

## Converting text (categorical labels) into numbers using LableEncoder

```python
attributes.head()
```

```python
result.head()
```

```python
le = LabelEncoder()

for column in attributes.columns:
    attributes[column] = le.fit_transform(attributes[column])  # converting each row of the column from text to number

result = le.fit_transform(result)
```

```python
attributes.head()
```

```python
result
```

## Now creating the model

```python
model = DecisionTreeClassifier()
model.fit(attributes, result)
```

```python
le.classes_
```

## Plotting the tree

```python
plt.figure(figsize=(7,10))
tree.plot_tree(model, feature_names= attributes.columns, class_names= le.classes_, filled= True)
plt.title('Decision Tree')
plt.show()
```
